# Para genera N archivos con M números aleatorios

El primer `for loop` se usa para que al momento de ingresar la cantidad de archivos que se desea, cada uno sea creado y al momento de entrar al segundo `for loop`, se les coloque un número M de números aleatorios. Para llevar estos `files.dat` con números aleatorios, se escogió tomar el primer número aleatorio dado por el output de `random.sh`.

`generar_files_i_N.sh`

```bash

#! /bin/bash

for i in $(seq 0 ${1}); do
        touch rand_${i}_${2}.dat
        for j in $(seq 1 ${2}); do
                out=$(./random.sh)
                first_number=$(echo "$out" | grep -E '^[0-9]+$' | head -n 1)
                echo "$first_number" >> rand_${i}_${2}.dat
        done
done



`seq 0 ${1}` : Genera una secuencia de enteros de 0 a el primer argumento del bash script

`seq 0 ${2}`: Genera una secuencia de enteros de 0 a el segundo argumento del bash script

`touch rand_${i}_${2}.dat` : Generar un archivo vacio con el nombre dado por el iterador i y el argumento que indica cuantos números aleatorios deben haber en cada archivo

`out=$(./random.sh)` : Se corre y se guarda en la variable `out` el output del bash script `random.sh`

`first_number=$(echo "$out" | grep -E '^[0-9]+$' | head -n 1)` : Se filta el contenido de `out` de modo que se obtenda solo el primer número aleatorio del `output` de `random.sh`. `echo "$out"` es para redirigir esa variable por el *pipe* a la función `grep` y con ella filtrar el contenido. `grep -E` es para interpretar patrones como "extended regular expressions" y el filtro `'^[0-9]+$'` es para que solo tome valores numéricos que se encuentra en el `output`. Se hace un *pipe* y se le pasa a la función `head` con el argumento `-n` para obtener los primeras lineas y se fija con el 1 que se busca solo la primera línea del `output`.

`echo "$first_number" >> rand_${i}_${2}.dat` : Se dirigue el resultado anterior al archivo corriente y se concatena con cada iteración del ciclo.

# Para generar un nuevo archivo con los datos de todos los demás

`buscar_data.sh`

```bash

#! /bin/bash

numero_archivos=$(ls rand_*.dat | wc -l) # Para saber cuantos archivos con números aleatorios hay

paste $(ls rand_*.dat | sort -V) > rand_${numero_archivos}.dat # Para poner las columnas de cada archivo una al lado de la otra en un nuevo archivo en orden
