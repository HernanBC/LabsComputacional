## a)

Para la realización de este problema hay que relacionar una distribución de probabilidad uniforme con una no uniforme, tal que:

$$z = \int_0^{x(z)} p(x'){\rm{d}} x'.$$

donde:

$p(x)$ : Representa la distribución de probabilidad no uniforme, que para este caso es la distribución de probabilidad de Boltzman.

$$\Rightarrow p(E) = \beta e^{-\beta E }$$

Según los visto en clases, la expresión inicial toma la forma:

$$\Rightarrow\ E = -\frac{\ln(1-z)}{\beta}.$$

## b)

Usando $\beta = 1$, se generan los siguiente números $z \in [0,1) $ en distribución uniforme y luego se transforman a una distribución de Boltzman

````python
import numpy as np

z = np.random.uniform(0.0,1.0,50000)

E = -1.0*np.log(1-z)
````

## c)

Histograma con numpy

````python
hist , bin_edges = np.histogram(E)
display(hist, bin_edges)
````

## d)

Histograma con matplotlib

````python
import matplotlib.pyplot as plt

valores_extra = np.arange(np.min(E), np.max(E),0.001) # Valores de puntos extras de interes para obtener su respectiva probabilidad
prob_density_func = (1.0)* np.exp(-(1.0) * valores_extra) # Función de densidad de probabilidad exponencial

fig, axes = plt.subplots(1,2, figsize = (10,5))

axes[0].hist(E,bins='auto')
axes[0].set_title("Histograma solicitado " + r"$\beta = 1.0$")
axes[0].set_xlabel("Energía")
axes[0].set_ylabel("Frecuencia de aparición")

axes[1].hist(E,bins='auto', density = True, label = "Histograma normalizado" )
axes[1].set_title("Comportamiento exponencial"  + r"$\beta = 1.0$")
axes[1].plot(valores_extra,prob_density_func, label = "FDP exponencial") # Función de distribución de probablilidad exponencial (FDP)
axes[1].set_xlabel("Energía")
axes[1].set_ylabel("Densidad de probabilidad")
axes[1].legend()
````

### Análisis

Al normalizar el histograma y juntarlo con una gráfica de la distribución de probablididad exponencial, es posible observar que las probabilidades de los datos obtenidos tras la conversión de números de distribución uniforme a distribución exponencial; sigue la distribución de densidad de probabilidades exponencial. Lo anterior es claro al ver como se reduce muy rapidamente la densidad probabilidad obtenida de los datos al alejarse de $x=0$ y como concuerda con el comportamiento teoríco de una gráfica exponencial de exponenete negativo. 

Lo anterior cumple debido a como se generon los números aleatorios. Estos fueron pesados con cierta probabilidad, antes de ser generados, dando como resultado que los números que poseían mayores probabilidades de ser obtenidos, se repetian un mayor número de veces y la forma es que fueron distribuidas dichas probablilidades concuerda con la de la distribución exponencial de probabilidad. El acto de pesar la generación de los números aleatorios fue realizado de forma indirecta al utilizar una función que mapea números con una distribución uniforme a números con una distribución exponencial.

# Problema 2

````python
# Función que va a ser integrada
def math_func(x): # x es la variable independiente de la función matemática a integrar
	return np.exp(-x**2)
````

````python
# Función para realizar integrales por el método de Monte Carlo de punto medio, devuelve el resultado de la integral con su respectivo error

def int_MonteCarlo_pm(integrando, x_inf, x_sup, can_numeros):
	x_values = np.random.uniform(low = x_inf, high= x_sup, size = can_numeros) # Se generan los números aleatorios dentro del intervalo de integración y para la cantidad deseada
	var_f = (1.0/can_numeros)*np.sum(integrando(x_values)**2.0) - ((1.0/can_numeros)*np.sum(integrando(x_values)))**2.0
	return ((x_sup-x_inf)/can_numeros)*np.sum(integrando(x_values)) , (x_sup-x_inf)*np.sqrt(var_f/can_numeros)# Se retorna el valor de la integral según la forma de Monte Carlo de punto medio, aprovechando el código compilado de numpy  de modo que no se utilizan for loops y los números pseudoaleatorios
````

````python
# Resultado de la evaluación de la integral por medio del método de Monte Carlo para el punto medio.
resultado , error = int_MonteCarlo_pm(math_func,0.0,1.0,100000)
display(resultado, error)
````

````python
# Gráfico del error respecto a la cantidad de números pseudoaleatorios

cantidad_numeros = np.linspace(10,100000,100, dtype='int') # Cantidad de números pseudoaletorios a usar para cada punto
errores = np.zeros(len(cantidad_numeros)) # Array donde se van a guardar los respectivos valores del error

for i in range(len(cantidad_numeros)):
	errores[i] = int_MonteCarlo_pm(math_func,0.0,1.0,cantidad_numeros[i])[1] # Guardando el error de implementación respectivo a la cantidad de números

fig, ax = plt.subplots(1,1, figsize = (8,5)) # Comienzo de la gráfica, se define el tamaño

ax.plot(cantidad_numeros, errores, '-o', ms = 2, label = "Error de implementación") # Se grafica el error de implementación
ax.plot(cantidad_numeros, 1/np.sqrt(cantidad_numeros), '-', ms = 2, label = "Error teórico " + r"$\frac{C}{\sqrt{N}}$") # Se gráfica el error teórico

ax.set_title("Error de implementación y teórico del uso del Método de Monte Carlo punto medio para integrales") # Título del gŕafico
ax.set_xscale("log", base=10) # Colocando escala logarítmica base 10 en eje x
ax.set_xlabel("Orden de Magnitud de la Cantidad de Números aleatorios") # Título del eje x
ax.set_yscale("log", base=10) # Colocando escala logarítmica base 10 en eje y
ax.set_ylabel(r"$log_{10}(Error)$") # Título del eje y
ax.grid(which='both') # Se coloca una grilla en la gráfica que incluye cuales marcar de los ejes (mayores y menores) tendran línea de grilla, ambos en este caso
ax.legend() # Se pide que se coloque la leyenda definida en cada función para graficar
````

## Análisis

Como se observa en en la gráfica anterior, las curvas del error de la implementación realizada y el error teórico tienen el mismo comportamiendo lineal al usar una escala logarítmica en ambos ejes (base = 10), por lo que sí se obtuvo el mismo comportamiento de los errores. Como fue comentado durante las lecciones, la convergencia de los resultados por medio de este método es muy lenta y requiere de grandes cantidades de números pseudoaleatorios para alcanzar resultados precisos; tal y como se evidencia en el gráfico anterior.
